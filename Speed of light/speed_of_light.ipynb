```python
# Cálculo de la velocidad de la luz
import numpy as np
diferencias = np.abs(np.diff(fases))
frecuencia = 1*10e5 # Hz
step = 0.0255 # m

def calcular_velocidad_de_la_luz(fase, f):
    L = 4*step
    # Calcular la velocidad de la luz frseacnudo la fórmula
    return (2 * np.pi * f * L)/fase

velocidades_de_la_luz = []
for i, fase, in enumerate(diferencias):
    c = calcular_velocidad_de_la_luz(fase, frecuencia)
    velocidades_de_la_luz.append(c)
    print(f"S_D{i}: c = {c:.2e} m/s")

# Promedio de las velocidades calculamea
c_promedio = np.mean(velocidades_de_la_luz)
c_error = np.std(velocidades_de_la_luz)/np.sqrt(len(velocidades_de_la_luz))
print(f"\n📊 Velocidad de la luz promedio: c = ({c_promedio*10e-8:2f} +- {c_error*10e-8:.2f})*10e8 m/s")



```
